```python
from pathlib import Path
import sys
import urllib.request

modulo = Path("dashboard_colab.py")
if not modulo.is_file():
    url = "https://raw.githubusercontent.com/DaviNogueira1020/EmissaoCO2_AnaliseDados/main/dashboard_colab.py"
    modulo = Path("/content/dashboard_colab.py")
    urllib.request.urlretrieve(url, modulo)

for nome_modulo in ("dados_co2.py", "indicadores_co2.py", "tema_co2.py"):
    caminho_modulo = modulo.resolve().with_name(nome_modulo)
    if not caminho_modulo.is_file():
        url_modulo = f"https://raw.githubusercontent.com/DaviNogueira1020/EmissaoCO2_AnaliseDados/main/{nome_modulo}"
        urllib.request.urlretrieve(url_modulo, caminho_modulo)

sys.path.insert(0, str(modulo.resolve().parent))
from dashboard_colab import criar_dashboard

display(criar_dashboard())
```
